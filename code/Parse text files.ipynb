```python
df3.rename(columns={'author': 'authors', 'journal': 'publication', 'pmid': 'pubmed_id'}, inplace=True)
df3.loc[:, 'source'] = 'scopus (1) (1).bib'
```

```python
columns = ['authors', 'title', 'abstract', 'year', 'publication', 'doi', 'pubmed_id', 'source']
all3 = pd.concat([df[columns] for df in [df1, df2, df3]], axis=0)
all3.head()
```

```python
all3['title'].isna().sum()
```

```python
all3[all3['title'].isna()]
```

```python
all3 = all3[~all3['title'].isna()]
```

```python
all3['title_extract'] = all3['title'].apply(extract_letters)
len(all3), all3['title_extract'].nunique()
```

```python
all3.isna().mean()
```

```python
all3.to_excel(r'..\all3_before_deduplication.xlsx', index=False)
```

```python
def first_non_na(s):
    # returns the first non-NA value in the Series, else np.nan
    s = s.dropna()
    return s.iloc[0] if len(s) else np.nan

dedup = all3.groupby('title_extract', as_index=False).agg(first_non_na)
len(dedup)
```

```python
dedup.isna().mean()
```
