```python
import os
import pandas as pd
from collections import defaultdict
from itertools import zip_longest
```

```python
def make_size_table(root):
    if not root.endswith('\\'):
        root += '\\'

    table = defaultdict(list)
    for dir, subdirs, files in os.walk(root):
        for file in files:
            if not file.endswith('.pdf'):
                continue
            path = os.path.join(dir, file)
            size = os.path.getsize(path)
            table[size].append(path.removeprefix(root))

    return table
```

```python
paul_table = make_size_table(r'C:\Users\israe\OneDrive\Documents\post doc\kfir\covid\Paul\Included studies')
```

```python
paul_table
```

```python
not_paul_table = make_size_table(r'C:\Users\israe\OneDrive\Documents\post doc\kfir\covid\Included studies')
```

```python
sizes = set(paul_table.keys()) | set(not_paul_table.keys())
rows = []
for size in sizes:
    paul_files = paul_table[size]
    not_paul_files = not_paul_table[size]
    comment = ''
    if not paul_files:
        comment = 'Not Paul only'
    elif not not_paul_files:
        comment = 'Paul only'
    elif len(paul_files) > 1 or len(not_paul_files) > 1:
        comment = 'Not unique'

    for paul, not_paul in zip_longest(paul_files, not_paul_files):
        rows.append({'size': size, 'Paul': paul, 'not Paul': not_paul, 'comment': comment})

df = pd.DataFrame(rows)
df
```
