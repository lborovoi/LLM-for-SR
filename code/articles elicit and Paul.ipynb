```python
import pandas as pd
```

```python
IN_PAUL = R'C:\Users\israe\OneDrive\Documents\post doc\kfir\covid\Elicit\COVID19_Studies_Papers_Paul.xlsx'
IN_ELICIT = r'C:\Users\israe\OneDrive\Documents\post doc\kfir\covid\Elicit\papers_tables\Elicit - extract-results-review-selected-until -2022.csv'
OUT_FILE = R'C:\Users\israe\OneDrive\Documents\post doc\kfir\covid\Elicit\merged.xlsx'
```

```python
paul_df = pd.read_excel(IN_PAUL)
paul_df.head()
```

```python
elicit_df = pd.read_csv(IN_ELICIT)
elicit_df.head()
```

```python
len(paul_df), len(paul_df.Title.unique())
```

```python
len(elicit_df), len(set(elicit_df.Title))
```

```python
elicit_df = elicit_df[elicit_df.Title.notna()].copy()
len(elicit_df)
```

```python
paul_df['in_paul'] = True
paul_df['in_elicit'] = paul_df.Title.isin(elicit_df.Title)
paul_df.in_elicit.sum()
```

```python
elicit_df['in_elicit'] = True
elicit_df['in_paul'] = elicit_df.Title.isin(paul_df.Title)
elicit_df.in_paul.sum()
```

```python
combined_df = pd.concat([
    elicit_df,
    paul_df[~paul_df.in_elicit]
], axis=0)
combined_df.sort_values('Title', inplace=True)
combined_df
```

```python
combined_df.to_excel(OUT_FILE, index=False)
```
